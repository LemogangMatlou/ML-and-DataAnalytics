# Teen Phone Addiction Dataset - Exploratory Data Analysis (EDA)

This notebook explores the *Teen Phone Addiction Dataset* with 3000 records and 25 columns.
We will look at:
- Dataset overview & summary statistics
- Correlations between features
- Distribution of Addiction Levels
- Key relationships (phone usage vs sleep, academics, gender, purpose)


```

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
file_path = "teen_phone_addiction_dataset.csv"
df = pd.read_csv(file_path)

df.shape, df.head()
```

## Summary Statistics

```
df.describe(include='all')
```

## Correlation Heatmap

```

plt.figure(figsize=(12,8))
sns.heatmap(df.corr(numeric_only=True), cmap="coolwarm", center=0)
plt.title("Correlation Heatmap of Numeric Features")
plt.show()
```

## Distribution of Addiction Level

```

plt.figure(figsize=(7,5))
sns.histplot(df["Addiction_Level"], kde=True, bins=30)
plt.title("Distribution of Addiction Level")
plt.xlabel("Addiction Level")
plt.ylabel("Count")
plt.show()
```

## Daily Usage vs Sleep Hours

```

plt.figure(figsize=(7,5))
sns.scatterplot(data=df, x="Daily_Usage_Hours", y="Sleep_Hours", hue="Addiction_Level", palette="viridis")
plt.title("Daily Usage vs Sleep Hours (colored by Addiction Level)")
plt.show()
```

## Addiction Level by Gender

```

plt.figure(figsize=(7,5))
sns.boxplot(data=df, x="Gender", y="Addiction_Level")
plt.title("Addiction Level by Gender")
plt.show()
```

## Phone Usage Purpose Distribution

```

plt.figure(figsize=(8,5))
sns.countplot(data=df, x="Phone_Usage_Purpose", order=df["Phone_Usage_Purpose"].value_counts().index)
plt.xticks(rotation=45)
plt.title("Distribution of Phone Usage Purpose")
plt.show()
```
